```python
import requests
import pandas as pd

# 1. جلب البيانات من GitHub API
url = "https://api.github.com/search/repositories?q=stars:>10000&sort=stars"
response = requests.get(url)
data = response.json()

# 2. تحويل البيانات إلى DataFrame
repos = data['items']
df = pd.DataFrame(repos)

# 3. اختيار الأعمدة واستخراج البيانات المتداخلة (owner و license)
df['owner_name'] = df['owner'].apply(lambda x: x['login'] if isinstance(x, dict) else None)
df['license_name'] = df['license'].apply(lambda x: x['name'] if isinstance(x, dict) and x else None)

columns_to_keep = [
    'id', 'name', 'full_name', 'owner_name', 'html_url',
    'description', 'stargazers_count', 'watchers_count',
    'language', 'forks_count', 'open_issues_count',
    'license_name', 'created_at', 'updated_at'
]

df_cleaned = df[columns_to_keep].copy()

# 4. معالجة القيم المفقودة والتكرارات
df_cleaned.drop_duplicates(subset=['id'], inplace=True)
df_cleaned['language'] = df_cleaned['language'].fillna('Unknown')
df_cleaned['license_name'] = df_cleaned['license_name'].fillna('No License')

# 5. تحويل التواريخ
df_cleaned['created_at'] = pd.to_datetime(df_cleaned['created_at'])
df_cleaned['updated_at'] = pd.to_datetime(df_cleaned['updated_at'])

# 6. حفظ الملف كـ CSV
df_cleaned.to_csv('github_projects.csv', index=False)
print("تم تنفيذ الخطوة الأولى وحفظ github_projects.csv بنجاح!")
```

```python
import sqlite3

# 1. إنشاء قاعدة بيانات SQLite باسم github_data.db وربطها
conn = sqlite3.connect('github_data.db')

# 2. نقل البيانات من Dataframe إلى جدول اسمه Repositories
df_cleaned.to_sql('Repositories', conn, if_exists='replace', index=False)

# 3. تشغيل استعلامات SQL مطلوبة للتحليل
print("--- 1. أثرى 5 مشاريع من حيث النجوم (Sorting & Limiting) ---")
query1 = """
SELECT name, language, stargazers_count
FROM Repositories
ORDER BY stargazers_count DESC
LIMIT 5;
"""
print(pd.read_sql_query(query1, conn))

print("\n--- 2. متوسط عدد النجوم وعدد المشاريع لكل لغة (Grouping & Aggregation) ---")
query2 = """
SELECT language, COUNT(*) as repo_count, AVG(stargazers_count) as avg_stars
FROM Repositories
GROUP BY language
HAVING COUNT(*) > 1
ORDER BY repo_count DESC;
"""
print(pd.read_sql_query(query2, conn))

print("\n--- 3. تصفية المشاريع بلغة Python أو JavaScript وتتجاوز 20,000 نجمة (Filtering & Logic) ---")
query3 = """
SELECT name, language, stargazers_count
FROM Repositories
WHERE (language = 'Python' OR language = 'JavaScript')
  AND stargazers_count > 20000;
"""
print(pd.read_sql_query(query3, conn))

# إغلاق الاتصال
conn.close()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# ضبط مظهر الرسم البياني
plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
fig, axes = plt.subplots(1, 2, figsize=(15, 6))

# 1. رسم بار تشارت لأكثر اللغات تكراراً (Bar Chart)
top_languages = df_cleaned['language'].value_counts().head(7)
axes[0].bar(top_languages.index, top_languages.values, color='#1f77b4')
axes[0].set_title('Top 7 Programming Languages', fontsize=14, fontweight='bold')
axes[0].set_xlabel('Language', fontsize=12)
axes[0].set_ylabel('Number of Repositories', fontsize=12)
axes[0].tick_params(axis='x', rotation=45)

# 2. رسم دائري لتوزيع اللغات (Pie Chart)
axes[1].pie(top_languages.values, labels=top_languages.index, autopct='%1.1f%%', startangle=140)
axes[1].set_title('Language Distribution', fontsize=14, fontweight='bold')

plt.tight_layout()
plt.show()
```
